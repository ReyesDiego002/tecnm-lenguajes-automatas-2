````python
%%writefile api_sesion21.h
/* api_sesion21.h - contrato de la Sesion 21
 * Cuádruplos y temporales únicos
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION21_H
#define API_SESION21_H

void nuevo_temporal(char *dst, int cap, int *contador);
void emitir_cuadruplo(char *dst, int cap, const char *op, const char *a1, const char *a2, const char *res);

#endif /* API_SESION21_H */
````

````python
%%writefile artesano_sesion21_c.c
/* artesano_sesion21_c.c - Fase 1: El Artesano (sin IA)
 * Cuádruplos y temporales únicos.
 * No lleva main(), el main es del harness.
 */
#include "api_sesion21.h"
#include <stdio.h>
#include <string.h>

void nuevo_temporal(char *dst, int cap, int *contador) {
    if (dst == NULL || cap <= 0) return;
    if (contador == NULL) { dst[0] = '\0'; return; }

    // primero uso el numero y despues lo subo
    snprintf(dst, (size_t)cap, "_t%d", *contador);
    (*contador)++;
}

void emitir_cuadruplo(char *dst, int cap, const char *op, const char *a1, const char *a2, const char *res) {
    if (dst == NULL || cap <= 0) return;

    // si alguno viene NULL pongo "_"
    if (op == NULL)  op = "_";
    if (a1 == NULL)  a1 = "_";
    if (a2 == NULL)  a2 = "_";
    if (res == NULL) res = "_";

    snprintf(dst, (size_t)cap, "(%s, %s, %s, %s)", op, a1, a2, res);
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
En Lenguajes y Autómatas II empezamos con código intermedio: el código de tres direcciones (TAC). Cada instrucción se guarda como un cuádruplo `(op, arg1, arg2, resultado)`. Para guardar resultados parciales se usan variables temporales `_t0`, `_t1`, `_t2`... que no se pueden repetir.
Las funciones no imprimen nada: escriben el texto en un buffer `dst` de tamaño `cap` que les pasan.
Es en C (C11) y se compila con `-Wall -Wextra -Werror`, sin warnings. Solo librerías estándar, sin `malloc`, sin variables globales y sin `main()`, porque el harness ya trae uno. Las firmas están en `api_sesion21.h` y no se cambian.

**R - Rol:**
Que la IA actúe como ingeniero de software de sistemas y compiladores que sabe generar código intermedio y manejar buffers de texto de forma segura.

**E - Especificación:**
1. `void nuevo_temporal(char *dst, int cap, int *contador);`
   - Escribe `_t` seguido del valor actual del contador (por ejemplo, `_t0`).
   - **Después** de escribir, incrementa el contador. Si el contador vale 0, escribe `_t0` y lo deja en 1.
   - Esto es importante: si el contador no se incrementa, todos los temporales salen iguales.

2. `void emitir_cuadruplo(char *dst, int cap, const char *op, const char *a1, const char *a2, const char *res);`
   - Escribe exactamente `(op, a1, a2, res)`, con coma y un espacio entre cada campo.
   - Si algún argumento es NULL se escribe `_` en su lugar.

- En las dos: nunca escribir más de `cap-1` caracteres (usar `snprintf`), siempre terminar en `'\0'`, y si `dst` es NULL o `cap <= 0` no hacer nada.

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| nuevo_temporal con contador 0 | "_t0" (contador queda en 1) |
| nuevo_temporal con contador 7 | "_t7" (contador queda en 8) |
| ("+", "a", "b", "_t0") | "(+, a, b, _t0)" |
| ("*", "_t0", "c", "_t1") | "(*, _t0, c, _t1)" |
| ("=", "a", NULL, "x") | "(=, a, _, x)" |
| ("goto", "L1", NULL, NULL) | "(goto, L1, _, _)" |

**O - Output:**
Solo el código de `ia_sesion21_c.c`, corto, sin `main()` y sin explicación larga. Al final una línea con la complejidad.

**Lo que revisé antes de compilar:**
Que no cambiara las firmas, que no trajera `main()` ni globales, que incrementara el contador después de usarlo y que usara `snprintf` con `cap`. Revisé a mano el contador después de llamar, porque el harness no lo checa.

````python
%%writefile ia_sesion21_c.c
/* ia_sesion21_c.c - Fase 2: El Cientifico
 * Codigo que me dio la IA (misma firma, sin main).
 */
#include "api_sesion21.h"
#include <stdio.h>

// si viene NULL regresa "_"
static const char *o_guion(const char *s) {
    return s ? s : "_";
}

void nuevo_temporal(char *dst, int cap, int *contador) {
    if (!dst || cap <= 0) return;
    if (!contador) { dst[0] = '\0'; return; }
    snprintf(dst, (size_t)cap, "_t%d", *contador);
    *contador += 1;   // se usa y DESPUES se incrementa
}

void emitir_cuadruplo(char *dst, int cap, const char *op, const char *a1, const char *a2, const char *res) {
    if (!dst || cap <= 0) return;
    snprintf(dst, (size_t)cap, "(%s, %s, %s, %s)",
             o_guion(op), o_guion(a1), o_guion(a2), o_guion(res));
}

// Complejidad: O(L) en las dos, L = largo del texto (limitado por cap)
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion21_c.c
/* harness_sesion21_c.c - auditoria automatica de la Sesion 21
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion21_c.c artesano_sesion21_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion21.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 21 - Cuadruplos y temporales unicos ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int c = 0; char b[32] = ""; nuevo_temporal(b, 32, &c); verificar_str("nuevo_temporal(_t0)", "_t0", b); }
    { int c = 7; char b[32] = ""; nuevo_temporal(b, 32, &c); verificar_str("nuevo_temporal(_t7)", "_t7", b); }
    { char b[64] = ""; emitir_cuadruplo(b, 64, "+", "a", "b", "_t0"); verificar_str_largo("cuadruplo de a + b", "(+, a, b, _t0)", b); }
    { char b[64] = ""; emitir_cuadruplo(b, 64, "*", "_t0", "c", "_t1"); verificar_str_largo("cuadruplo de _t0 * c", "(*, _t0, c, _t1)", b); }
    { char b[64] = ""; emitir_cuadruplo(b, 64, "=", "a", NULL, "x"); verificar_str("asignacion simple", "(=, a, _, x)", b); }
    { char b[64] = ""; emitir_cuadruplo(b, 64, "goto", "L1", NULL, NULL); verificar_str_largo("goto sin operandos", "(goto, L1, _, _)", b); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion21.md" "\n");
    return 0;
}
````
