````python
%%writefile api_sesion23.h
/* api_sesion23.h - contrato de la Sesion 23
 * Etiquetas y saltos incondicionales
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION23_H
#define API_SESION23_H

void nueva_etiqueta(char *dst, int cap, int *contador);
void emitir_flujo(char *dst, int cap, const char *etiqueta, int es_salto);

#endif /* API_SESION23_H */
````

````python
%%writefile artesano_sesion23_c.c
/* artesano_sesion23_c.c - Fase 1: El Artesano (sin IA)
 * Etiquetas y saltos incondicionales.
 * No lleva main(), el main es del harness.
 */
#include "api_sesion23.h"
#include <stdio.h>
#include <string.h>

void nueva_etiqueta(char *dst, int cap, int *contador) {
    if (dst == NULL || cap <= 0) return;
    if (contador == NULL) { dst[0] = '\0'; return; }

    // uso el numero y luego lo subo (igual que los temporales)
    snprintf(dst, (size_t)cap, "L%d", *contador);
    (*contador)++;
}

void emitir_flujo(char *dst, int cap, const char *etiqueta, int es_salto) {
    if (dst == NULL || cap <= 0) return;

    // sin etiqueta no hay nada que escribir
    if (etiqueta == NULL) { dst[0] = '\0'; return; }

    if (es_salto != 0) {
        snprintf(dst, (size_t)cap, "GOTO %s", etiqueta); // el salto
    } else {
        snprintf(dst, (size_t)cap, "%s:", etiqueta);     // la marca de destino
    }
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
En Lenguajes y Autómatas II estamos con código intermedio, ahora con saltos. Para moverse a otra parte del programa se usan etiquetas `L0`, `L1`, `L2`... Hay dos formas de usarlas: la marca de destino (`L2:`), que dice "aquí es", y el salto (`GOTO L2`), que dice "vete para allá".
Las funciones no imprimen: escriben en un buffer `dst` de tamaño `cap`.
Es en C (C11) y se compila con `-Wall -Wextra -Werror`, sin warnings. Solo librerías estándar, sin `malloc`, sin globales y sin `main()`, porque el harness ya trae uno. Las firmas están en `api_sesion23.h` y no se cambian.

**R - Rol:**
Que la IA actúe como ingeniero de software de sistemas y compiladores que sabe generar código intermedio con etiquetas y saltos.

**E - Especificación:**
1. `void nueva_etiqueta(char *dst, int cap, int *contador);`
   - Escribe `L` + el valor actual del contador (por ejemplo `L0`) y **después** incrementa el contador.
   - La primera etiqueta es `L0` y el contador queda en 1. Si el contador vale 10, escribe `L10`.

2. `void emitir_flujo(char *dst, int cap, const char *etiqueta, int es_salto);`
   - Si `es_salto` es **distinto de cero** (1, 2, -1, cualquier valor que no sea 0) escribe `GOTO etiqueta`, con un espacio.
   - Si `es_salto` es 0 escribe `etiqueta:`, con dos puntos pegados y sin espacio.
   - Si `etiqueta` es NULL escribe cadena vacía y no truena.

- En las dos: usar `snprintf` con `cap`, nunca escribir fuera del buffer, y si `dst` es NULL o `cap <= 0` no hacer nada.

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| nueva_etiqueta con contador 0 | "L0" |
| nueva_etiqueta con contador 5 | "L5" |
| emitir_flujo("L2", 1) | "GOTO L2" |
| emitir_flujo("L2", 0) | "L2:" |
| emitir_flujo(NULL, 1) | "" |
| nueva_etiqueta con contador 10 | "L10" |

**O - Output:**
Solo el código de `ia_sesion23_c.c`, corto, sin `main()` y sin explicación larga. Al final una línea con la complejidad.

**Lo que revisé antes de compilar:**
Que no cambiara las firmas, que no trajera `main()`, que incrementara el contador después de usarlo y que tomara cualquier valor distinto de 0 como salto, no solo el 1.

````python
%%writefile ia_sesion23_c.c
/* ia_sesion23_c.c - Fase 2: El Cientifico
 * Codigo que me dio la IA (misma firma, sin main).
 */
#include "api_sesion23.h"
#include <stdio.h>

void nueva_etiqueta(char *dst, int cap, int *contador) {
    if (!dst || cap <= 0) return;
    if (!contador) { dst[0] = '\0'; return; }
    snprintf(dst, (size_t)cap, "L%d", *contador);
    *contador += 1;   // post-incremento: primero L0
}

void emitir_flujo(char *dst, int cap, const char *etiqueta, int es_salto) {
    if (!dst || cap <= 0) return;
    if (!etiqueta) { dst[0] = '\0'; return; }
    // cualquier valor distinto de 0 cuenta como salto
    snprintf(dst, (size_t)cap, es_salto ? "GOTO %s" : "%s:", etiqueta);
}

// Complejidad: O(L) en las dos, L = largo del texto (limitado por cap)
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion23_c.c
/* harness_sesion23_c.c - auditoria automatica de la Sesion 23
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion23_c.c artesano_sesion23_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion23.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 23 - Etiquetas y saltos incondicionales ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int c = 0; char b[32] = ""; nueva_etiqueta(b, 32, &c); verificar_str("nueva_etiqueta(L0)", "L0", b); }
    { int c = 5; char b[32] = ""; nueva_etiqueta(b, 32, &c); verificar_str("nueva_etiqueta(L5)", "L5", b); }
    { char b[32] = ""; emitir_flujo(b, 32, "L2", 1); verificar_str("emitir_flujo(GOTO L2)", "GOTO L2", b); }
    { char b[32] = ""; emitir_flujo(b, 32, "L2", 0); verificar_str("emitir_flujo(marca L2)", "L2:", b); }
    { char b[32] = ""; emitir_flujo(b, 32, NULL, 1); verificar_str("emitir_flujo(NULL)", "", b); }
    { int c = 10; char b[32] = ""; nueva_etiqueta(b, 32, &c); verificar_str("nueva_etiqueta(L10)", "L10", b); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion23.md" "\n");
    return 0;
}
````
