## Fase 1 · El Artesano (15 min, sin IA)

Escribe a mano la implementación en `artesano_sesion18_c.c`. El archivo ya trae la

````python
%%writefile api_sesion18.h
/* api_sesion18.h - contrato de la Sesion 18
 * Tamaño de arreglo y límites
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION18_H
#define API_SESION18_H

long bytes_arreglo(const char *tipo, int elementos);
int indice_en_rango(int indice, int elementos);

#endif /* API_SESION18_H */
````

````python
%%writefile artesano_sesion18_c.c
/* artesano_sesion18_c.c - Fase 1: El Artesano (sin IA)
 * Tamaño de arreglo y límites.
 * No lleva main(), el main es del harness.
 */
#include "api_sesion18.h"
#include <stdio.h>
#include <string.h>

// cuantos bytes mide cada tipo, -1 si no existe
static int ancho(const char *tipo) {
    if (strcmp(tipo, "char") == 0)    return 1;
    if (strcmp(tipo, "short") == 0)   return 2;
    if (strcmp(tipo, "int") == 0)     return 4;
    if (strcmp(tipo, "float") == 0)   return 4;
    if (strcmp(tipo, "double") == 0)  return 8;
    if (strcmp(tipo, "puntero") == 0) return 8;
    return -1;
}

long bytes_arreglo(const char *tipo, int elementos) {
    if (tipo == NULL || elementos < 0) return -1;

    int a = ancho(tipo);
    if (a == -1) return -1;

    // lo paso a long antes de multiplicar para que no se desborde
    return (long)elementos * a;
}

int indice_en_rango(int indice, int elementos) {
    // va de 0 hasta elementos-1, el "elementos" ya no cuenta
    if (indice >= 0 && indice < elementos) return 1;
    return 0;
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
En Lenguajes y Autómatas II estamos viendo arreglos. El compilador tiene que calcular cuántos bytes reservar para un arreglo (elementos × tamaño del tipo) y revisar que los índices no se salgan del arreglo.
Es en C (C11) y se compila con `-Wall -Wextra -Werror`, sin warnings. Solo librerías estándar y sin `main()`, porque el harness ya trae uno. Las firmas están en `api_sesion18.h` y no se cambian. Ojo: `bytes_arreglo` regresa `long` pero `elementos` es `int`.

**R - Rol:**
Que la IA actúe como ingeniero de software de sistemas y compiladores que sabe cómo se calcula la memoria de un arreglo y cómo se validan los límites.

**E - Especificación:**
1. `long bytes_arreglo(const char *tipo, int elementos);`
   - Anchos: `char=1`, `short=2`, `int=4`, `float=4`, `double=8`, `puntero=8` (en español, `"puntero"`).
   - Regresa `elementos * ancho`.
   - La multiplicación se tiene que hacer en `long` (castear antes de multiplicar), para que con arreglos grandes no se desborde el `int`.
   - Cero elementos regresa 0.
   - Elementos negativos, tipo NULL o tipo desconocido regresan -1.

2. `int indice_en_rango(int indice, int elementos);`
   - Regresa 1 si `0 <= indice < elementos`, y 0 si no.
   - El índice más alto válido es `elementos - 1`, nunca `elementos`.
   - Los índices negativos regresan 0.

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| bytes_arreglo("int", 10) | 40 |
| bytes_arreglo("double", 3) | 24 |
| bytes_arreglo("int", -4) | -1 |
| indice_en_rango(0, 5) | 1 |
| indice_en_rango(5, 5) | 0 |
| indice_en_rango(-1, 5) | 0 |

**O - Output:**
Solo el código de `ia_sesion18_c.c`, corto, sin `main()` y sin explicación larga. Al final una línea con la complejidad.

**Lo que revisé antes de compilar:**
Que no cambiara las firmas, que no trajera `main()`, que usara `<` y no `<=` en el índice y que hiciera la multiplicación en `long`. Lo probé con `double[300000000]`.

````python
%%writefile ia_sesion18_c.c
/* ia_sesion18_c.c - Fase 2: El Cientifico
 * Codigo que me dio la IA (misma firma, sin main).
 */
#include "api_sesion18.h"
#include <string.h>
#include <stddef.h>

long bytes_arreglo(const char *tipo, int elementos) {
    const char *nombres[] = {"char", "short", "int", "float", "double", "puntero"};
    const int anchos[]    = { 1,      2,       4,     4,       8,        8 };

    if (!tipo || elementos < 0) return -1;

    for (int i = 0; i < 6; i++) {
        if (strcmp(tipo, nombres[i]) == 0) {
            return (long)elementos * anchos[i]; // en long para no desbordar
        }
    }
    return -1; // tipo desconocido
}

int indice_en_rango(int indice, int elementos) {
    return (indice >= 0 && indice < elementos) ? 1 : 0;
}

// Complejidad: bytes_arreglo O(1) (6 tipos fijos), indice_en_rango O(1)
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion18_c.c
/* harness_sesion18_c.c - auditoria automatica de la Sesion 18
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion18_c.c artesano_sesion18_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion18.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 18 - Tamano de arreglo y limites ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { long r = bytes_arreglo("int", 10); verificar_int("int[10]", (long)(40), (long)(r)); }
    { long r = bytes_arreglo("double", 3); verificar_int("double[3]", (long)(24), (long)(r)); }
    { long r = bytes_arreglo("int", -4); verificar_int("int[-4]", (long)(-1), (long)(r)); }
    { int r = indice_en_rango(0, 5); verificar_int("indice 0 de 5", (long)(1), (long)(r)); }
    { int r = indice_en_rango(5, 5); verificar_int("indice 5 de 5", (long)(0), (long)(r)); }
    { int r = indice_en_rango(-1, 5); verificar_int("indice negativo", (long)(0), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion18.md" "\n");
    return 0;
}
````
