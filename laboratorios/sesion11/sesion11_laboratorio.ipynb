Escribe a mano la implementación en `artesano_sesion11_c.c`. El archivo ya trae la

````python
%%writefile api_sesion11.h
/* api_sesion11.h - contrato de la Sesion 11
 * Tabla hash con encadenamiento
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION11_H
#define API_SESION11_H

typedef struct Simbolo {
    char nombre[16];
    int tipo;
    int offset;
    struct Simbolo *sig;
} Simbolo;

unsigned int hash_djb2(const char *clave);
int insertar_simbolo(Simbolo *tabla[], unsigned int cubetas, const char *nombre, int tipo, int offset);

#endif /* API_SESION11_H */
````

````python
%%writefile artesano_sesion11_c.c
/* artesano_sesion11_c.c - Fase 1: El Artesano (sin IA)
 * Tabla hash con encadenamiento.
 *
 * Implementa lo que declara api_sesion11.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion11.h"
#include <stdio.h>
#include <string.h>
#include <stdlib.h>

unsigned int hash_djb2(const char *clave) {
    unsigned int h = 5381;

    if (clave == NULL) {
        return h;
    }

    // por cada letra: h = h * 33 + letra
    while (*clave != '\0') {
        h = h * 33 + (unsigned char)*clave;
        clave++;
    }
    return h;
}

int insertar_simbolo(Simbolo *tabla[], unsigned int cubetas, const char *nombre, int tipo, int offset) {
    // primero reviso que me manden cosas validas
    if (tabla == NULL || nombre == NULL || cubetas == 0) {
        return -1;
    }

    // el nombre solo cabe en 15 letras + '\0', asi que lo corto aqui
    char nom[16];
    strncpy(nom, nombre, sizeof(nom) - 1);
    nom[sizeof(nom) - 1] = '\0';

    unsigned int pos = hash_djb2(nom) % cubetas;

    // recorro la cadena de esa cubeta para ver si ya existe
    Simbolo *actual = tabla[pos];
    while (actual != NULL) {
        if (strcmp(actual->nombre, nom) == 0) {
            return 0; // ya estaba, no lo meto otra vez
        }
        actual = actual->sig;
    }

    // no existe, creo el nodo nuevo
    Simbolo *nuevo = malloc(sizeof(Simbolo));
    if (nuevo == NULL) {
        return -1; // no hubo memoria
    }
    strcpy(nuevo->nombre, nom);
    nuevo->tipo = tipo;
    nuevo->offset = offset;

    // lo pongo al inicio de la cadena, asi no piso a nadie
    nuevo->sig = tabla[pos];
    tabla[pos] = nuevo;

    return 1;
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
Estoy en la materia de Lenguajes y Autómatas II y ahorita estamos viendo la tabla de símbolos, que es donde el compilador guarda los nombres de las variables. Tengo que hacerla con una tabla hash con encadenamiento (cada cubeta es una lista enlazada).
El código es en C (C11) y se compila con GCC con `-Wall -Wextra -Werror`, o sea que si sale cualquier warning ya no compila. Solo puedo usar librerías estándar. Mi archivo se compila junto con un harness que ya tiene su `main()`, así que yo no debo poner `main()`. Las firmas y el `struct Simbolo` vienen en `api_sesion11.h` y no los puedo cambiar.
Algo importante: el campo `nombre` del struct es `char nombre[16]`, o sea que solo caben 15 letras más el `'\0'`.

**R - Rol:**
Le pedí a la IA que se comportara como un ingeniero de software de sistemas y compiladores, que sepa de tablas hash, listas enlazadas y manejo de memoria en C.

**E - Especificación:**
Tiene que hacer dos funciones:

1. `unsigned int hash_djb2(const char *clave);`
   - Empieza en 5381 y por cada carácter hace `hash = hash * 33 + c`.
   - `hash_djb2("")` debe dar 5381 y `hash_djb2("abc")` debe dar 193485963.

2. `int insertar_simbolo(Simbolo *tabla[], unsigned int cubetas, const char *nombre, int tipo, int offset);`
   - La cubeta se saca con `hash_djb2(nombre) % cubetas`.
   - Si el nombre ya existe, regresa 0 y no crea nodo.
   - Si es nuevo, crea un nodo con `malloc`, guarda nombre, tipo y offset, lo encadena en la cubeta y regresa 1.
   - Si dos nombres caen en la misma cubeta, se encadenan. Nunca se sobrescribe uno que ya estaba.
   - Si `tabla` es NULL, `nombre` es NULL o `cubetas` es 0, regresa -1.
   - El nombre se copia sin pasarse de los 16 bytes del campo (nada de `strcpy` directo con el nombre que llega).

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| hash_djb2("") | 5381 |
| hash_djb2("abc") | 193485963 |
| insertar "x" dos veces | la segunda regresa 0 |
| "alpha" y "beta" con 1 cubeta | 2 nodos en la cadena |
| offset de "beta" en la cadena | 4 |
| insertar con tabla NULL | -1 |

**O - Output:**
Solo el código del archivo `ia_sesion11_c.c`, sin `main()` y sin explicación larga. Al final una línea con la complejidad de cada función.

**Lo que revisé antes de compilar:**
Que no cambiara las firmas ni el struct, que no trajera `main()` y que no usara `strcpy` directo con el nombre. Luego lo corrí con el harness y con un nombre largo para ver que no se saliera de la memoria.

````python
%%writefile ia_sesion11_c.c
/* ia_sesion11_c.c - Fase 2: El Cientifico
 *
 * Codigo que me devolvio la IA, con la MISMA firma
 * que declara api_sesion11.h. No trae main().
 */
#include "api_sesion11.h"
#include <stdlib.h>
#include <string.h>

unsigned int hash_djb2(const char *clave) {
    unsigned int hash = 5381;
    if (clave == NULL) return hash;
    for (const unsigned char *p = (const unsigned char *)clave; *p; p++) {
        hash = ((hash << 5) + hash) + *p; /* hash * 33 + c */
    }
    return hash;
}

int insertar_simbolo(Simbolo *tabla[], unsigned int cubetas, const char *nombre, int tipo, int offset) {
    if (tabla == NULL || nombre == NULL || cubetas == 0) {
        return -1;
    }

    /* copia segura: el campo nombre solo tiene 16 bytes */
    char clave[sizeof(((Simbolo *)0)->nombre)];
    strncpy(clave, nombre, sizeof(clave) - 1);
    clave[sizeof(clave) - 1] = '\0';

    unsigned int idx = hash_djb2(clave) % cubetas;

    for (Simbolo *s = tabla[idx]; s != NULL; s = s->sig) {
        if (strcmp(s->nombre, clave) == 0) {
            return 0; /* duplicado */
        }
    }

    Simbolo *nuevo = (Simbolo *)malloc(sizeof(Simbolo));
    if (nuevo == NULL) return -1;

    memcpy(nuevo->nombre, clave, sizeof(clave));
    nuevo->tipo = tipo;
    nuevo->offset = offset;
    nuevo->sig = tabla[idx]; /* insercion al frente de la cadena */
    tabla[idx] = nuevo;
    return 1;
}

/* Complejidad: hash O(L); insertar O(L + k), k = largo de la cadena de la cubeta. */
````

## Fase 3 · El Auditor (10 min)

````python
%%writefile harness_sesion11_c.c
/* harness_sesion11_c.c - auditoria automatica de la Sesion 11
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion11_c.c artesano_sesion11_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion11.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 11 - Tabla hash con encadenamiento ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { unsigned int r = hash_djb2(""); verificar_int("hash_djb2(\"\")", (long)(5381), (long)(r)); }
    { unsigned int r = hash_djb2("abc"); verificar_int("hash_djb2(\"abc\")", (long)(193485963), (long)(r)); }
    { Simbolo *t[8] = {0}; insertar_simbolo(t, 8, "x", 1, 0); int r = insertar_simbolo(t, 8, "x", 2, 8); verificar_int("insertar x dos veces", (long)(0), (long)(r)); }
    { Simbolo *t[1] = {0}; insertar_simbolo(t, 1, "alpha", 1, 0); insertar_simbolo(t, 1, "beta", 2, 4); int r = 0; for (Simbolo *p = t[0]; p != NULL; p = p->sig) r++; verificar_int("dos claves, una cubeta", (long)(2), (long)(r)); }
    { Simbolo *t[1] = {0}; insertar_simbolo(t, 1, "alpha", 1, 0); insertar_simbolo(t, 1, "beta", 2, 4); int r = -1; for (Simbolo *p = t[0]; p != NULL; p = p->sig) { if (strcmp(p->nombre, "beta") == 0) r = p->offset; } verificar_int("offset de beta en cadena", (long)(4), (long)(r)); }
    { int r = insertar_simbolo(NULL, 8, "x", 1, 0); verificar_int("insertar en tabla NULL", (long)(-1), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion11.md" "\n");
    return 0;
}
````
