````python
%%writefile api_sesion26.h
/* api_sesion26.h - contrato de la Sesion 26
 * Llamadas, parámetros y retorno
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION26_H
#define API_SESION26_H

int emitir_llamada(char lineas[][64], int max, int *n, const char *funcion, const char *a1, const char *a2, const char *resultado);
void emitir_return(char *dst, int cap, const char *valor);

#endif /* API_SESION26_H */
````

````python
%%writefile artesano_sesion26_c.c
/* artesano_sesion26_c.c - Fase 1: El Artesano (sin IA)
 * Llamadas, parámetros y retorno.
 * No lleva main(), el main es del harness.
 */
#include "api_sesion26.h"
#include <stdio.h>
#include <string.h>

int emitir_llamada(char lineas[][64], int max, int *n, const char *funcion, const char *a1, const char *a2, const char *resultado) {
    if (lineas == NULL || n == NULL || funcion == NULL) return 0;

    int escritas = 0; // lineas que meti en esta llamada
    int params = 0;   // cuantos PARAM salieron de verdad

    // un PARAM por cada argumento que no sea NULL (primero a1, luego a2)
    if (a1 != NULL && *n < max) {
        snprintf(lineas[*n], 64, "PARAM %s", a1);
        (*n)++; escritas++; params++;
    }
    if (a2 != NULL && *n < max) {
        snprintf(lineas[*n], 64, "PARAM %s", a2);
        (*n)++; escritas++; params++;
    }

    // el CALL lleva el numero real de PARAM
    if (*n < max) {
        if (resultado != NULL)
            snprintf(lineas[*n], 64, "%s = CALL %s, %d", resultado, funcion, params);
        else
            snprintf(lineas[*n], 64, "CALL %s, %d", funcion, params); // no asigna
        (*n)++; escritas++;
    }
    return escritas;
}

void emitir_return(char *dst, int cap, const char *valor) {
    if (dst == NULL || cap <= 0) return;

    if (valor != NULL) snprintf(dst, (size_t)cap, "RETURN %s", valor);
    else               snprintf(dst, (size_t)cap, "RETURN"); // funcion void
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
En Lenguajes y Autómatas II estamos traduciendo llamadas a funciones a código intermedio (TAC). Una llamada como `_t0 = f(x, y)` se convierte en: primero un `PARAM` por cada argumento, luego el `CALL` con el nombre de la función y cuántos parámetros lleva. Para regresar un valor se usa `RETURN`.
`emitir_llamada` agrega líneas al arreglo `lineas` desde la posición `*n` y va subiendo `*n`. `emitir_return` escribe en un buffer `dst` de tamaño `cap`.
Es en C (C11) y se compila con `-Wall -Wextra -Werror`, sin warnings. Solo librerías estándar y sin `main()`, porque el harness ya trae uno. Las firmas están en `api_sesion26.h` y no se cambian.

**R - Rol:**
Que la IA actúe como ingeniero de software de sistemas y compiladores que sabe generar TAC para llamadas a funciones y retornos.

**E - Especificación:**
1. `int emitir_llamada(char lineas[][64], int max, int *n, const char *funcion, const char *a1, const char *a2, const char *resultado);`
   - Revisa `a1` y `a2` **cada uno por separado**: cada argumento que no sea NULL genera una línea `PARAM arg`, en orden a1 y luego a2.
   - Si `a1` es NULL pero `a2` no, igual se escribe `PARAM a2` (no hay que detenerse en el primer NULL).
   - Después escribe el CALL con la cantidad **real** de PARAM que salieron:
     - Con resultado: `resultado = CALL funcion, k`
     - Si `resultado` es NULL: `CALL funcion, k` (sin asignar)
   - Regresa cuántas líneas agregó.
   - Nunca escribir más de `max` líneas. Si `lineas`, `n` o `funcion` son NULL, regresa 0.

2. `void emitir_return(char *dst, int cap, const char *valor);`
   - Si hay valor: `RETURN valor`. Si `valor` es NULL: solo `RETURN`.
   - Usar `snprintf` con `cap`.

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| f("x", "y", "_t0"): línea 0 | "PARAM x" |
| f("x", "y", "_t0"): línea 1 | "PARAM y" |
| f("x", "y", "_t0"): línea 2 | "_t0 = CALL f, 2" |
| g("x") → _t0 y luego f("_t0") → _t1: línea 3 | "_t1 = CALL f, 1" |
| p(NULL, NULL, "_t0"): línea 0 | "_t0 = CALL p, 0" |
| emitir_return("_t0") | "RETURN _t0" |

**O - Output:**
Solo el código de `ia_sesion26_c.c`, corto, sin `main()` y sin explicación larga. Al final una línea con la complejidad.

**Lo que revisé antes de compilar:**
Que no cambiara las firmas, que no trajera `main()`, que revisara a1 y a2 por separado y que el número del CALL fuera el de PARAM reales. Lo probé con `a1 = NULL` y `a2 = "y"`.

````python
%%writefile ia_sesion26_c.c
/* ia_sesion26_c.c - Fase 2: El Cientifico
 * Codigo que me dio la IA (misma firma, sin main).
 */
#include "api_sesion26.h"
#include <stdio.h>
#include <stddef.h>

int emitir_llamada(char lineas[][64], int max, int *n, const char *funcion, const char *a1, const char *a2, const char *resultado) {
    if (!lineas || !n || !funcion) return 0;

    const char *args[2] = { a1, a2 };
    int antes = *n, params = 0;

    // cada argumento no NULL es un PARAM (aunque a1 sea NULL, a2 se revisa)
    for (int i = 0; i < 2; i++) {
        if (args[i] && *n < max) {
            snprintf(lineas[(*n)++], 64, "PARAM %s", args[i]);
            params++;
        }
    }

    if (*n < max) {
        if (resultado) snprintf(lineas[(*n)++], 64, "%s = CALL %s, %d", resultado, funcion, params);
        else           snprintf(lineas[(*n)++], 64, "CALL %s, %d", funcion, params);
    }
    return *n - antes;   // lineas agregadas
}

void emitir_return(char *dst, int cap, const char *valor) {
    if (!dst || cap <= 0) return;
    if (valor) snprintf(dst, (size_t)cap, "RETURN %s", valor);
    else       snprintf(dst, (size_t)cap, "RETURN");
}

// Complejidad: O(1) (maximo 3 lineas por llamada)
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,

````python
%%writefile harness_sesion26_c.c
/* harness_sesion26_c.c - auditoria automatica de la Sesion 26
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion26_c.c artesano_sesion26_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion26.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 26 - Llamadas, parametros y retorno ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { char l[16][64] = {{0}}; int n = 0; emitir_llamada(l, 16, &n, "f", "x", "y", "_t0"); verificar_str("PARAM x", "PARAM x", l[0]); }
    { char l[16][64] = {{0}}; int n = 0; emitir_llamada(l, 16, &n, "f", "x", "y", "_t0"); verificar_str("PARAM y", "PARAM y", l[1]); }
    { char l[16][64] = {{0}}; int n = 0; emitir_llamada(l, 16, &n, "f", "x", "y", "_t0"); verificar_str_largo("CALL con dos argumentos", "_t0 = CALL f, 2", l[2]); }
    { char l[16][64] = {{0}}; int n = 0; emitir_llamada(l, 16, &n, "g", "x", NULL, "_t0"); emitir_llamada(l, 16, &n, "f", "_t0", NULL, "_t1"); verificar_str_largo("llamada anidada f(g(x))", "_t1 = CALL f, 1", l[3]); }
    { char l[8][64] = {{0}}; int n = 0; emitir_llamada(l, 8, &n, "p", NULL, NULL, "_t0"); verificar_str_largo("CALL sin argumentos", "_t0 = CALL p, 0", l[0]); }
    { char b[32] = ""; emitir_return(b, 32, "_t0"); verificar_str("RETURN con valor", "RETURN _t0", b); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion26.md" "\n");
    return 0;
}
````
