````python
%%writefile api_sesion22.h
/* api_sesion22.h - contrato de la Sesion 22
 * Traductor de AST a TAC con precedencia
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION22_H
#define API_SESION22_H

typedef struct Nodo {
    char tipo;
    char op;
    char valor[16];
    struct Nodo *izq;
    struct Nodo *der;
} Nodo;

int generar_tac(const Nodo *raiz, char lineas[][64], int max);

#endif /* API_SESION22_H */
````

````python
%%writefile artesano_sesion22_c.c
/* artesano_sesion22_c.c - Fase 1: El Artesano (sin IA)
 * Traductor de AST a TAC con precedencia.
 * No lleva main(), el main es del harness.
 */
#include "api_sesion22.h"
#include <stdio.h>
#include <string.h>

/* Los temporales ya estan resueltos: usalos tal cual. */
typedef struct {
    char nombre[64][16];
    int usados;
} Pool;

const char *nuevo_temporal(Pool *p, int *contador) {
    if (p->usados >= 64) return "_t?";
    snprintf(p->nombre[p->usados], sizeof p->nombre[0], "_t%d", (*contador)++);
    return p->nombre[p->usados++];
}

// recorre el arbol en postorden y regresa el nombre donde quedo el resultado
static const char *recorrer(const Nodo *n, char lineas[][64], int max, int *num, Pool *p, int *cont) {
    if (n == NULL) return "_";

    // si es hoja no genera linea, solo regresa su nombre
    if (n->tipo == 'H') return n->valor;

    // primero los hijos (izq y der), despues el operador
    const char *izq = recorrer(n->izq, lineas, max, num, p, cont);
    const char *der = recorrer(n->der, lineas, max, num, p, cont);
    const char *t = nuevo_temporal(p, cont);

    // solo escribo si todavia hay lugar
    if (*num < max) {
        snprintf(lineas[*num], 64, "%s = %s %c %s", t, izq, n->op, der);
        (*num)++;
    }
    return t;
}

int generar_tac(const Nodo *raiz, char lineas[][64], int max) {
    if (raiz == NULL || lineas == NULL || max <= 0) return 0;

    Pool p;
    p.usados = 0;
    int contador = 0;
    int num = 0;

    recorrer(raiz, lineas, max, &num, &p, &contador);
    return num;
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
En Lenguajes y Autómatas II estamos generando código de tres direcciones (TAC) a partir de un árbol de expresión (AST). Cada nodo es una hoja (`tipo = 'H'`, con su nombre en `valor`) o un operador binario (`tipo = 'B'`, con `op` y sus hijos `izq` y `der`).
Hay que recorrer el árbol y escribir una línea TAC por cada operador en el arreglo `lineas`, sin perder la precedencia (lo que está más abajo en el árbol se calcula primero).
La práctica ya trae un `Pool` y la función `nuevo_temporal(Pool *p, int *contador)`, que da `_t0`, `_t1`, ... Hay que usarla, no inventar los nombres.
Es en C (C11) y se compila con `-Wall -Wextra -Werror`, sin warnings. Solo librerías estándar y sin `main()`, porque el harness ya trae uno. El struct `Nodo` y la firma están en `api_sesion22.h` y no se cambian.

**R - Rol:**
Que la IA actúe como ingeniero de software de sistemas y compiladores que sabe recorrer árboles y generar código intermedio.

**E - Especificación:**
`int generar_tac(const Nodo *raiz, char lineas[][64], int max);`
- Recorrido en **postorden**: primero el hijo izquierdo, luego el derecho y al final el operador.
- Una hoja no genera línea; solo regresa su nombre (`valor`).
- Cada operador pide un temporal nuevo con `nuevo_temporal` y escribe `"_tX = izq op der"` (con un espacio entre cada parte).
- Regresa cuántas líneas escribió.
- **Nunca** escribe más de `max` líneas (si ya no hay lugar, deja de escribir) ni más de 63 caracteres por línea (usar `snprintf` con 64).
- Si `raiz` es NULL, `lineas` es NULL o `max <= 0`, regresa 0.
- Incluir en el archivo el mismo `Pool` y `nuevo_temporal` que da la práctica.

Casos que debe pasar:
| Árbol | Línea revisada | Esperado |
| :--- | :--- | :--- |
| a + b * c | línea 0 | `_t0 = b * c` |
| a + b * c | línea 1 | `_t1 = a + _t0` |
| (a + b) * c | línea 1 | `_t1 = _t0 * c` |
| x * (y - z) + w | línea 2 | `_t2 = _t1 + w` |
| solo la hoja x | regresa | 0 |
| (a + b) * (c - d) | línea 2 | `_t2 = _t0 * _t1` |

**O - Output:**
Solo el código de `ia_sesion22_c.c`, corto, sin `main()` y sin explicación larga. Al final una línea con la complejidad.

**Lo que revisé antes de compilar:**
Que no cambiara la firma, que no trajera `main()`, que fuera postorden y que revisara `max` antes de escribir cada línea. Lo probé con `max = 1` en un árbol de 3 operadores.

````python
%%writefile ia_sesion22_c.c
/* ia_sesion22_c.c - Fase 2: El Cientifico
 * Codigo que me dio la IA (misma firma, sin main).
 */
#include "api_sesion22.h"
#include <stdio.h>

/* Pool y nuevo_temporal iguales a los que da la practica */
typedef struct {
    char nombre[64][16];
    int usados;
} Pool;

const char *nuevo_temporal(Pool *p, int *contador) {
    if (p->usados >= 64) return "_t?";
    snprintf(p->nombre[p->usados], sizeof p->nombre[0], "_t%d", (*contador)++);
    return p->nombre[p->usados++];
}

typedef struct {
    char (*lineas)[64];
    int max;
    int n;        // lineas escritas
    int cont;     // contador de temporales
    Pool pool;
} Gen;

static const char *postorden(const Nodo *nd, Gen *g) {
    if (!nd) return "_";
    if (nd->tipo == 'H') return nd->valor;          // hoja: no emite

    const char *a = postorden(nd->izq, g);          // hijos primero
    const char *b = postorden(nd->der, g);
    const char *t = nuevo_temporal(&g->pool, &g->cont);

    if (g->n < g->max) {                            // nunca mas de max lineas
        snprintf(g->lineas[g->n], 64, "%s = %s %c %s", t, a, nd->op, b);
        g->n++;
    }
    return t;
}

int generar_tac(const Nodo *raiz, char lineas[][64], int max) {
    if (!raiz || !lineas || max <= 0) return 0;
    Gen g;
    g.lineas = lineas;
    g.max = max;
    g.n = 0;
    g.cont = 0;
    g.pool.usados = 0;
    postorden(raiz, &g);
    return g.n;
}

// Complejidad: O(n) en tiempo, n = nodos del arbol; memoria O(h) de recursion, h = altura
````

````python
%%writefile harness_sesion22_c.c
/* harness_sesion22_c.c - auditoria automatica de la Sesion 22
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion22_c.c artesano_sesion22_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion22.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 22 - Traductor de AST a TAC con precedencia ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { Nodo a = {.tipo='H', .valor="a"}; Nodo b = {.tipo='H', .valor="b"}; Nodo c = {.tipo='H', .valor="c"}; Nodo m = {.tipo='B', .op='*', .izq=&b, .der=&c}; Nodo s = {.tipo='B', .op='+', .izq=&a, .der=&m}; char l[8][64] = {{0}}; generar_tac(&s, l, 8); verificar_str("a + b * c", "_t0 = b * c", l[0]); }
    { Nodo a = {.tipo='H', .valor="a"}; Nodo b = {.tipo='H', .valor="b"}; Nodo c = {.tipo='H', .valor="c"}; Nodo m = {.tipo='B', .op='*', .izq=&b, .der=&c}; Nodo s = {.tipo='B', .op='+', .izq=&a, .der=&m}; char l[8][64] = {{0}}; generar_tac(&s, l, 8); verificar_str_largo("a + b * c (2a linea)", "_t1 = a + _t0", l[1]); }
    { Nodo a = {.tipo='H', .valor="a"}; Nodo b = {.tipo='H', .valor="b"}; Nodo c = {.tipo='H', .valor="c"}; Nodo s = {.tipo='B', .op='+', .izq=&a, .der=&b}; Nodo m = {.tipo='B', .op='*', .izq=&s, .der=&c}; char l[8][64] = {{0}}; generar_tac(&m, l, 8); verificar_str_largo("(a + b) * c", "_t1 = _t0 * c", l[1]); }
    { Nodo x = {.tipo='H', .valor="x"}; Nodo y = {.tipo='H', .valor="y"}; Nodo z = {.tipo='H', .valor="z"}; Nodo w = {.tipo='H', .valor="w"}; Nodo r = {.tipo='B', .op='-', .izq=&y, .der=&z}; Nodo m = {.tipo='B', .op='*', .izq=&x, .der=&r}; Nodo s = {.tipo='B', .op='+', .izq=&m, .der=&w}; char l[8][64] = {{0}}; generar_tac(&s, l, 8); verificar_str_largo("x * (y - z) + w", "_t2 = _t1 + w", l[2]); }
    { Nodo a = {.tipo='H', .valor="x"}; char l[4][64] = {{0}}; int n = generar_tac(&a, l, 4); verificar_int("hoja sin operaciones", (long)(0), (long)(n)); }
    { Nodo a = {.tipo='H', .valor="a"}; Nodo b = {.tipo='H', .valor="b"}; Nodo c = {.tipo='H', .valor="c"}; Nodo d = {.tipo='H', .valor="d"}; Nodo s1 = {.tipo='B', .op='+', .izq=&a, .der=&b}; Nodo s2 = {.tipo='B', .op='-', .izq=&c, .der=&d}; Nodo m = {.tipo='B', .op='*', .izq=&s1, .der=&s2}; char l[8][64] = {{0}}; generar_tac(&m, l, 8); verificar_str_largo("(a + b) * (c - d)", "_t2 = _t0 * _t1", l[2]); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion22.md" "\n");
    return 0;
}
````
