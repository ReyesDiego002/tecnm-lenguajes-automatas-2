## Fase 1 · El Artesano (15 min, sin IA)

Escribe a mano la implementación en `artesano_sesion15_c.c`. El archivo ya trae la
firma y un `TODO`: edita el contenido de la celda `%%writefile` que está abajo (o el

````python
%%writefile api_sesion15.h
/* api_sesion15.h - contrato de la Sesion 15
 * Push y pop de ámbitos
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION15_H
#define API_SESION15_H

typedef struct Ambito {
    const char *nombre;
    struct Ambito *padre;
} Ambito;

Ambito *push_ambito(Ambito *actual, const char *nombre);
Ambito *pop_ambito(Ambito *actual);

#endif /* API_SESION15_H */
````

````python
%%writefile artesano_sesion15_c.c
/* artesano_sesion15_c.c - Fase 1: El Artesano (sin IA)
 * Push y pop de ámbitos.
 * No lleva main(), el main es del harness.
 */
#include "api_sesion15.h"
#include <stdio.h>
#include <string.h>
#include <stdlib.h>

Ambito *push_ambito(Ambito *actual, const char *nombre) {
    if (actual == NULL || nombre == NULL) return NULL;

    Ambito *nuevo = malloc(sizeof(Ambito));
    if (nuevo == NULL) return NULL; // no hubo memoria

    nuevo->nombre = nombre;
    nuevo->padre = actual; // el nuevo apunta al que estaba
    return nuevo;
}

Ambito *pop_ambito(Ambito *actual) {
    if (actual == NULL) return NULL;

    // si no tiene papa es la raiz, esa no la hice yo con malloc, no la toco
    if (actual->padre == NULL) return NULL;

    Ambito *papa = actual->padre; // lo guardo ANTES de liberar
    free(actual);
    return papa;
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
En Lenguajes y Autómatas II estamos viendo qué pasa con los bloques anidados. Cuando el programa abre una llave `{` se crea un ámbito nuevo (push), y cuando la cierra `}` ese ámbito se destruye y se regresa al de afuera (pop).
Cada ámbito solo tiene un `nombre` y un puntero a su `padre`. La raíz (global) la crea el harness en la pila (no con `malloc`) y su `padre` es NULL.
Es en C (C11) y se compila con `-Wall -Wextra -Werror`, sin warnings. Solo librerías estándar y sin `main()`, porque el harness ya trae uno. El struct y las firmas están en `api_sesion15.h` y no se cambian.

**R - Rol:**
Que la IA actúe como ingeniero de software de sistemas y compiladores que sabe manejar memoria dinámica (`malloc`/`free`) y pilas enlazadas.

**E - Especificación:**
1. `Ambito *push_ambito(Ambito *actual, const char *nombre);`
   - Reserva un ámbito nuevo con `malloc`, le pone el nombre que recibe y lo enlaza con `padre = actual`.
   - Regresa el ámbito nuevo.
   - Si `actual` o `nombre` son NULL (o falla `malloc`), regresa NULL.

2. `Ambito *pop_ambito(Ambito *actual);`
   - Guarda el padre, libera con `free` el ámbito actual y regresa el padre.
   - Si `actual` es NULL o es la raíz (`padre == NULL`), regresa NULL y **no** hace `free`, porque la raíz no se creó con `malloc`.
   - Solo libera el ámbito que recibe, nunca a los padres.

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| push(&raiz, "if") y ver su nombre | "if" |
| dos push y contar niveles | 3 |
| pop del hijo | regresa &raiz |
| pop sobre la raíz | NULL, sin tronar |
| push con padre NULL | NULL |
| dos push, un pop y contar niveles | 2 |

**O - Output:**
Solo el código de `ia_sesion15_c.c`, corto, sin `main()` y sin explicación larga. Al final una línea con la complejidad.

**Lo que revisé antes de compilar:**
Que no cambiara las firmas, que no trajera `main()`, que en `pop` guardara el padre antes del `free` y que no hiciera `free` a la raíz.

````python
%%writefile ia_sesion15_c.c
/* ia_sesion15_c.c - Fase 2: El Cientifico
 * Codigo que me dio la IA (misma firma, sin main).
 */
#include "api_sesion15.h"
#include <stdlib.h>

Ambito *push_ambito(Ambito *actual, const char *nombre) {
    if (!actual || !nombre) return NULL;
    Ambito *a = malloc(sizeof *a);
    if (!a) return NULL;
    a->nombre = nombre;
    a->padre = actual;
    return a;
}

Ambito *pop_ambito(Ambito *actual) {
    // la raiz (sin padre) no se libera porque no la creo push
    if (!actual || !actual->padre) return NULL;
    Ambito *padre = actual->padre;
    free(actual);
    return padre;
}

// Complejidad: push O(1), pop O(1)
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,

````python
%%writefile harness_sesion15_c.c
/* harness_sesion15_c.c - auditoria automatica de la Sesion 15
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion15_c.c artesano_sesion15_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion15.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str(const char *caso, const char *esperado,
                          const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12s obtenido: %-12s %s\n",
           probadas, caso, esperado, obtenido ? obtenido : "(NULL)",
           ok ? "OK" : "FALLA");
}

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 15 - Push y pop de ambitos ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { Ambito raiz = {"global", NULL}; Ambito *a = push_ambito(&raiz, "if"); const char *r = a ? a->nombre : "(nulo)"; pop_ambito(a); verificar_str("nombre del ambito nuevo", "if", r); }
    { Ambito raiz = {"global", NULL}; Ambito *a = push_ambito(&raiz, "if"); Ambito *b = push_ambito(a, "while"); int r = 0; for (Ambito *p = b; p != NULL; p = p->padre) r++; pop_ambito(b); pop_ambito(a); verificar_int("profundidad con dos push", (long)(3), (long)(r)); }
    { Ambito raiz = {"global", NULL}; Ambito *a = push_ambito(&raiz, "if"); int r = (pop_ambito(a) == &raiz); verificar_int("pop devuelve el padre", (long)(1), (long)(r)); }
    { Ambito raiz = {"global", NULL}; int r = (pop_ambito(&raiz) == NULL); verificar_int("pop sobre la raiz", (long)(1), (long)(r)); }
    { int r = (push_ambito(NULL, "x") == NULL); verificar_int("push con padre NULL", (long)(1), (long)(r)); }
    { Ambito raiz = {"global", NULL}; Ambito *a = push_ambito(&raiz, "if"); Ambito *b = push_ambito(a, "while"); Ambito *c = pop_ambito(b); int r = 0; for (Ambito *p = c; p != NULL; p = p->padre) r++; pop_ambito(c); verificar_int("profundidad tras pop", (long)(2), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion15.md" "\n");
    return 0;
}
````
