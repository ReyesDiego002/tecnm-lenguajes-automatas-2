````python
%%writefile api_sesion17.h
/* api_sesion17.h - contrato de la Sesion 17
 * Validación de retornos y rutas
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION17_H
#define API_SESION17_H

int diagnosticar_retorno(const char *tipo_funcion, const char *tipo_expresion, int rutas_totales, int rutas_con_return);

#endif /* API_SESION17_H */
````

````python
%%writefile artesano_sesion17_c.c
/* artesano_sesion17_c.c - Fase 1: El Artesano (sin IA)
 * Validación de retornos y rutas.
 * No lleva main(), el main es del harness.
 */
#include "api_sesion17.h"
#include <stdio.h>
#include <string.h>

// que tan "ancho" es cada tipo, void es 0 y -1 si no lo conozco
static int rango(const char *t) {
    if (strcmp(t, "void") == 0)   return 0;
    if (strcmp(t, "char") == 0)   return 1;
    if (strcmp(t, "int") == 0)    return 2;
    if (strcmp(t, "float") == 0)  return 3;
    if (strcmp(t, "double") == 0) return 4;
    return -1;
}

int diagnosticar_retorno(const char *tipo_funcion, const char *tipo_expresion, int rutas_totales, int rutas_con_return) {
    // 1) datos raros -> -1
    if (tipo_funcion == NULL || tipo_expresion == NULL) return -1;
    if (rutas_totales <= 0 || rutas_con_return < 0 || rutas_con_return > rutas_totales) return -1;

    int rf = rango(tipo_funcion);
    int re = rango(tipo_expresion);
    if (rf == -1 || re == -1) return -1; // tipo que no existe

    // 2) primero el tipo
    if (rf == 0 || re == 0) {
        if (rf == re) return 0; // void con void esta bien
        return 1;               // void devolviendo algo, o al reves
    }
    if (rf < re) return 1; // ej. int no puede recibir double

    // 3) luego las rutas: todas deben tener return
    if (rutas_con_return < rutas_totales) return 2;

    return 0; // todo bien
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
En Lenguajes y Autómatas II estamos viendo cómo el compilador revisa los `return`. Tiene que checar dos cosas: que lo que devuelves sea compatible con el tipo de la función, y que todos los caminos (rutas) de la función terminen en un `return`. Por ejemplo, un `if` sin `else` deja un camino abierto.
La función recibe el tipo de la función, el tipo de la expresión del return, cuántas rutas hay y cuántas de esas tienen `return`.
Es en C (C11) y se compila con `-Wall -Wextra -Werror`, sin warnings. Solo librerías estándar y sin `main()`, porque el harness ya trae uno. La firma está en `api_sesion17.h` y no se cambia.

**R - Rol:**
Que la IA actúe como ingeniero de software de sistemas y compiladores que sabe de comprobación semántica de tipos y de análisis de rutas de control.

**E - Especificación:**
`int diagnosticar_retorno(const char *tipo_funcion, const char *tipo_expresion, int rutas_totales, int rutas_con_return);`

Regresa un solo diagnóstico:
- **-1**: si algún tipo es NULL o no existe, si `rutas_totales <= 0`, si `rutas_con_return` es negativo o si hay más returns que rutas.
- **1** (error de tipo):
  - la función es `void` y la expresión no es `void` (o al revés);
  - la expresión es más ancha que la función (por ejemplo, `int` no acepta `double`).
  - Jerarquía de anchos: `char < int < float < double`. Se acepta cuando el tipo de la función es igual o más ancho que el de la expresión.
- **2**: si el tipo está bien pero `rutas_con_return < rutas_totales` (falta un return en alguna ruta).
- **0**: todo bien.

**Importante:** el tipo se revisa ANTES que las rutas. Si el tipo está mal y además falta una ruta, debe regresar 1, no 2.

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| ("int", "int", 2, 2) | 0 |
| ("void", "int", 1, 1) | 1 |
| ("int", "int", 2, 1) | 2 |
| ("double", "int", 1, 1) | 0 |
| ("int", "double", 1, 1) | 1 |
| ("int", "int", 0, 0) | -1 |

**O - Output:**
Solo el código de `ia_sesion17_c.c`, corto, sin `main()` y sin explicación larga. Al final una línea con la complejidad.

**Lo que revisé antes de compilar:**
Que no cambiara la firma, que no trajera `main()` y que revisara el tipo antes que las rutas. Lo probé con `("int", "double", 2, 1)`, que debe dar 1.

````python
%%writefile ia_sesion17_c.c
/* ia_sesion17_c.c - Fase 2: El Cientifico
 * Codigo que me dio la IA (misma firma, sin main).
 */
#include "api_sesion17.h"
#include <string.h>
#include <stddef.h>

// rango: void=0, char=1, int=2, float=3, double=4, desconocido=-1
static int rango(const char *t) {
    const char *tipos[] = {"void", "char", "int", "float", "double"};
    for (int i = 0; i < 5; i++) {
        if (strcmp(t, tipos[i]) == 0) return i;
    }
    return -1;
}

int diagnosticar_retorno(const char *tipo_funcion, const char *tipo_expresion, int rutas_totales, int rutas_con_return) {
    if (!tipo_funcion || !tipo_expresion) return -1;
    if (rutas_totales <= 0 || rutas_con_return < 0 || rutas_con_return > rutas_totales) return -1;

    int rf = rango(tipo_funcion), re = rango(tipo_expresion);
    if (rf < 0 || re < 0) return -1;

    // el tipo se revisa ANTES que las rutas
    if (rf == 0 || re == 0) return (rf == re) ? 0 : 1;
    if (re > rf) return 1;                         // se perderia informacion

    if (rutas_con_return < rutas_totales) return 2; // if sin else
    return 0;
}

// Complejidad: O(1)
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion17_c.c
/* harness_sesion17_c.c - auditoria automatica de la Sesion 17
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion17_c.c artesano_sesion17_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion17.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 17 - Validacion de retornos y rutas ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int r = diagnosticar_retorno("int", "int", 2, 2); verificar_int("int con int y dos rutas", (long)(0), (long)(r)); }
    { int r = diagnosticar_retorno("void", "int", 1, 1); verificar_int("void devuelve valor", (long)(1), (long)(r)); }
    { int r = diagnosticar_retorno("int", "int", 2, 1); verificar_int("ruta sin retorno", (long)(2), (long)(r)); }
    { int r = diagnosticar_retorno("double", "int", 1, 1); verificar_int("double acepta int", (long)(0), (long)(r)); }
    { int r = diagnosticar_retorno("int", "double", 1, 1); verificar_int("int no acepta double", (long)(1), (long)(r)); }
    { int r = diagnosticar_retorno("int", "int", 0, 0); verificar_int("cero rutas de ejecucion", (long)(-1), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion17.md" "\n");
    return 0;
}
````
