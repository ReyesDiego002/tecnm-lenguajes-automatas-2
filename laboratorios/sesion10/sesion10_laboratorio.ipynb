````python
%%writefile api_sesion10.h
/* api_sesion10.h - contrato de la Sesion 10
 * Mini-evaluador de expresiones (Hito 1)
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION10_H
#define API_SESION10_H

long evaluar_expresion(const char *fuente, int *error);

#endif /* API_SESION10_H */
````

````python
%%writefile artesano_sesion10_c.c
/* artesano_sesion10_c.c - Fase 1: El Artesano (sin IA)
 * Mini-evaluador de expresiones (Hito 1).
 *
 * Implementa lo que declara api_sesion10.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion10.h"
#include <stdio.h>
#include <string.h>

#include <ctype.h>

/* El analizador ya viene armado: completa las tres funciones marcadas. */
typedef struct { const char *s; int error; } Parser;

static void saltar(Parser *p) {
    while (*p->s == ' ') p->s++;
}

long expr(Parser *p);

long primaria(Parser *p) {
    long valor = 0;
    saltar(p);

    if (*p->s == '(') {
        p->s++;                 /* consumimos '(' */
        valor = expr(p);
        if (p->error != 0) return 0;
        saltar(p);
        if (*p->s != ')') {     /* falta el ')' */
            p->error = 2;
            return 0;
        }
        p->s++;                 /* consumimos ')' */
        return valor;
    }

    if (isdigit((unsigned char)*p->s)) {
        while (isdigit((unsigned char)*p->s)) {
            valor = valor * 10 + (*p->s - '0');
            p->s++;
        }
        return valor;
    }

    /* no es '(' ni digito: error de sintaxis */
    p->error = 2;
    return 0;
}

long termino(Parser *p) {
    long valor = primaria(p);

    while (p->error == 0) {
        saltar(p);
        char op = *p->s;
        if (op != '*' && op != '/') break;
        p->s++;                 /* consumimos el operador */

        long der = primaria(p);
        if (p->error != 0) return 0;

        if (op == '*') {
            valor = valor * der;
        } else {
            if (der == 0) {     /* division entre cero */
                p->error = 1;
                return 0;
            }
            valor = valor / der;
        }
    }
    return valor;
}

long expr(Parser *p) {
    long valor = termino(p);

    while (p->error == 0) {
        saltar(p);
        char op = *p->s;
        if (op != '+' && op != '-') break;
        p->s++;                 /* consumimos el operador */

        long der = termino(p);
        if (p->error != 0) return 0;

        if (op == '+') valor = valor + der;
        else           valor = valor - der;
    }
    return valor;
}

long evaluar_expresion(const char *fuente, int *error) {
    static int sin_error;
    if (error == NULL) error = &sin_error;
    *error = 0;
    if (fuente == NULL) { *error = 2; return 0; }
    Parser p = { fuente, 0 };
    long v = expr(&p);
    saltar(&p);
    if (p.error == 0 && *p.s != '\0') p.error = 2;
    *error = p.error;
    return p.error != 0 ? 0 : v;
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
Estoy haciendo el Hito 1 de la materia Lenguajes y Autómatas II: un mini-evaluador de expresiones aritméticas. El programa es en C (estándar C11) y se compila con GCC usando `-Wall -Wextra -Werror`, por eso cualquier advertencia cuenta como error. No puedo usar bibliotecas que no sean estándar (solo `stdio.h`, `string.h`, `ctype.h`). Mi archivo se compila junto con un harness que ya tiene su propio `main()`, así que mi código no debe tener `main()`. La firma está en `api_sesion10.h` y no se puede cambiar. El evaluador debe ser un analizador descendente recursivo con tres funciones: `primaria`, `termino` y `expr`.

**R - Rol:**
Le pido a la IA que actúe como ingeniero de software de sistemas y compiladores, que sepa escribir un analizador descendente recursivo a partir de una gramática y que respete la precedencia y la asociatividad de los operadores.

**E - Especificación:**
Implementar la función:
`long evaluar_expresion(const char *fuente, int *error);`

Gramática que debe seguir:
- `expr     -> termino (('+' | '-') termino)*`
- `termino  -> primaria (('*' | '/') primaria)*`
- `primaria -> NUMERO | '(' expr ')'`

Reglas:
- Solo enteros positivos y los operadores `+ - * /`. Sin variables ni menos unario.
- La multiplicación y la división van antes que la suma y la resta.
- Los operadores se asocian a la izquierda: `10-2-3` debe dar 5, no 11.
- Los paréntesis y los espacios se respetan y no cambian el resultado.
- `*error = 0` si todo sale bien.
- `*error = 1` si hay división entre cero (también si el cero sale de una operación, como `10/(5-5)`).
- `*error = 2` si la sintaxis es inválida: falta un número, falta un `)`, sobra texto al final o la cadena está vacía.
- Si hay error, la función devuelve 0.
- Si `error` es NULL no debe tronar.

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| "2+3*4" | 14 |
| "(2+3)*4" | 20 |
| "10/0" | error = 1 |
| "2+" | error = 2 |
| "  7 " | 7 |
| "5-10" | -5 |

**O - Output:**
Solo el código del archivo `ia_sesion10_c.c` (completo, incluyendo `evaluar_expresion`), sin `main()` y sin explicaciones largas. Al final una línea con la complejidad.

**Verificación antes de compilar:**
Revisé que la IA no cambiara la firma, que no metiera `main()` y que las funciones de la gramática estuvieran con ciclos `while` y no con recursión por la derecha. Después lo compilé con el harness para ver si salen los 6 casos en OK, y lo probé con casos extra como `10-2-3`, `8/2/2` y `10/(5-5)`.

````python
%%writefile ia_sesion10_c.c
/* ia_sesion10_c.c - Fase 2: El Cientifico
 *
 * Codigo devuelto por la IA, con la MISMA firma
 * que declara api_sesion10.h. No trae main().
 */
#include "api_sesion10.h"
#include <stddef.h>
#include <ctype.h>

/* Gramatica:
 *   expr     -> termino (('+' | '-') termino)*
 *   termino  -> primaria (('*' | '/') primaria)*
 *   primaria -> NUMERO | '(' expr ')'
 */
typedef struct { const char *s; int error; } Parser;

static void saltar(Parser *p) {
    while (*p->s == ' ') p->s++;
}

static long expr(Parser *p);

static long primaria(Parser *p) {
    long v = 0;
    saltar(p);
    if (*p->s == '(') {
        p->s++;
        v = expr(p);
        if (p->error) return 0;
        saltar(p);
        if (*p->s != ')') { p->error = 2; return 0; }
        p->s++;
        return v;
    }
    if (!isdigit((unsigned char)*p->s)) { p->error = 2; return 0; }
    while (isdigit((unsigned char)*p->s)) {
        v = v * 10 + (*p->s - '0');
        p->s++;
    }
    return v;
}

static long termino(Parser *p) {
    long v = primaria(p);
    while (!p->error) {
        saltar(p);
        char op = *p->s;
        if (op != '*' && op != '/') break;
        p->s++;
        long d = primaria(p);
        if (p->error) return 0;
        if (op == '/' && d == 0) { p->error = 1; return 0; }
        v = (op == '*') ? v * d : v / d;
    }
    return v;
}

static long expr(Parser *p) {
    long v = termino(p);
    while (!p->error) {
        saltar(p);
        char op = *p->s;
        if (op != '+' && op != '-') break;
        p->s++;
        long d = termino(p);
        if (p->error) return 0;
        v = (op == '+') ? v + d : v - d;
    }
    return v;
}

long evaluar_expresion(const char *fuente, int *error) {
    static int sin_error;
    if (error == NULL) error = &sin_error;
    *error = 0;
    if (fuente == NULL) { *error = 2; return 0; }
    Parser p = { fuente, 0 };
    long v = expr(&p);
    saltar(&p);
    if (p.error == 0 && *p.s != '\0') p.error = 2;   /* basura al final */
    *error = p.error;
    return p.error != 0 ? 0 : v;
}

/* Complejidad: O(n) en tiempo, n = largo de la cadena. Memoria O(k), k = profundidad de parentesis (recursion). */
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion10_c.c
/* harness_sesion10_c.c - auditoria automatica de la Sesion 10
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion10_c.c artesano_sesion10_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion10.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 10 - Mini-evaluador de expresiones (Hito 1) ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int e=0; long r=evaluar_expresion("2+3*4",&e); (void)e; verificar_int("\"2+3*4\"", (long)(14), (long)(r)); }
    { int e=0; long r=evaluar_expresion("(2+3)*4",&e); (void)e; verificar_int("\"(2+3)*4\"", (long)(20), (long)(r)); }
    { int e=0; long r=evaluar_expresion("10/0",&e); (void)r; verificar_int("\"10/0\" -> error 1", (long)(1), (long)(e)); }
    { int e=0; long r=evaluar_expresion("2+",&e); (void)r; verificar_int("\"2+\" -> error 2", (long)(2), (long)(e)); }
    { int e=0; long r=evaluar_expresion("  7 ",&e); (void)e; verificar_int("\"  7 \"", (long)(7), (long)(r)); }
    { int e=0; long r=evaluar_expresion("5-10",&e); (void)e; verificar_int("\"5-10\"", (long)(-5), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion10.md" "\n");
    return 0;
}
````
