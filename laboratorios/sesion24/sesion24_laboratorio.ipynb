````python
%%writefile artesano_sesion24_c.c
/* artesano_sesion24_c.c - Fase 1: El Artesano (sin IA)
 * Saltos condicionales con cortocircuito.
 * No lleva main(), el main es del harness.
 */
#include "api_sesion24.h"
#include <stdio.h>
#include <string.h>

int emitir_condicion(char lineas[][64], int max, const char *a, const char *b, char op, const char *etiqueta) {
    if (lineas == NULL || max <= 0 || a == NULL || etiqueta == NULL) return 0;

    // AND salta si es falso, OR salta si es verdadero
    const char *salto;
    if (op == '&')      salto = "IFFALSE";
    else if (op == '|') salto = "IFTRUE";
    else return 0; // operador que no conozco, no escribo nada

    int n = 0;

    // primer operando
    snprintf(lineas[n], 64, "%s %s GOTO %s", salto, a, etiqueta);
    n++;

    // segundo operando, solo si hay y si cabe
    if (b != NULL && n < max) {
        snprintf(lineas[n], 64, "%s %s GOTO %s", salto, b, etiqueta);
        n++;
    }
    return n;
}
````

### Fase 2 · El Científico — Contrato C-R-E-O

**C - Contexto:**
En Lenguajes y Autómatas II estamos traduciendo `if` a código intermedio. Como el procesador solo va instrucción por instrucción, un `if (a && b)` se convierte en saltos: si algo no se cumple, brinca a la etiqueta de salida y se salta el bloque. Esto se llama cortocircuito: con AND, si `a` es falso ya no importa `b`; con OR, si `a` es verdadero ya no importa `b`.
La función escribe las líneas en un arreglo `lineas` con espacio para `max` líneas de 64 caracteres.
Es en C (C11) y se compila con `-Wall -Wextra -Werror`, sin warnings. Solo librerías estándar y sin `main()`, porque el harness ya trae uno. La firma está en `api_sesion24.h` y no se cambia.

**R - Rol:**
Que la IA actúe como ingeniero de software de sistemas y compiladores que sabe generar saltos condicionales con cortocircuito.

**E - Especificación:**
`int emitir_condicion(char lineas[][64], int max, const char *a, const char *b, char op, const char *etiqueta);`
- Si `op` es `'&'` (AND), cada salto es `IFFALSE operando GOTO etiqueta`.
- Si `op` es `'|'` (OR), cada salto es `IFTRUE operando GOTO etiqueta`.
- **Cualquier otro operador** es inválido: regresa 0 y no escribe nada (no hay que tratarlo como OR ni como AND).
- Primero el salto de `a` y luego el de `b`, cada uno en su propia línea.
- Si `b` es NULL, solo se escribe el salto de `a`. No inventar una comparación.
- Regresa cuántas líneas escribió. Nunca más de `max`.
- Si `lineas`, `a` o `etiqueta` son NULL, o `max <= 0`, regresa 0.

Casos que debe pasar:
| Caso | Esperado |
| :--- | :--- |
| ("a", "b", '&', "L9") línea 0 | "IFFALSE a GOTO L9" |
| ("a", "b", '&', "L9") línea 1 | "IFFALSE b GOTO L9" |
| ("a", "b", '&', "L9") regresa | 2 |
| ("a", "b", '|', "L9") línea 0 | "IFTRUE a GOTO L9" |
| ("a", "b", '|', "L9") línea 1 | "IFTRUE b GOTO L9" |
| ("a", NULL, '&', "L0") línea 0 | "IFFALSE a GOTO L0" |

**O - Output:**
Solo el código de `ia_sesion24_c.c`, corto, sin `main()` y sin explicación larga. Al final una línea con la complejidad.

**Lo que revisé antes de compilar:**
Que no cambiara la firma, que no trajera `main()`, que un operador desconocido regresara 0 y que revisara `max` antes del segundo salto.

````python
%%writefile ia_sesion24_c.c
/* ia_sesion24_c.c - Fase 2: El Cientifico
 * Codigo que me dio la IA (misma firma, sin main).
 */
#include "api_sesion24.h"
#include <stdio.h>
#include <stddef.h>

int emitir_condicion(char lineas[][64], int max, const char *a, const char *b, char op, const char *etiqueta) {
    if (!lineas || max <= 0 || !a || !etiqueta) return 0;

    const char *instr = NULL;
    if (op == '&') instr = "IFFALSE";   // AND: si es falso ya no sigue
    if (op == '|') instr = "IFTRUE";    // OR: si es verdadero ya no sigue
    if (!instr) return 0;               // operador invalido

    const char *ops[2] = { a, b };
    int n = 0;
    for (int i = 0; i < 2 && n < max; i++) {
        if (!ops[i]) break;             // b NULL: un solo salto
        snprintf(lineas[n], 64, "%s %s GOTO %s", instr, ops[i], etiqueta);
        n++;
    }
    return n;
}

// Complejidad: O(1) (maximo 2 lineas)
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:

````python
%%writefile harness_sesion24_c.c
/* harness_sesion24_c.c - auditoria automatica de la Sesion 24
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion24_c.c artesano_sesion24_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion24.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_str_largo(const char *caso, const char *esperado,
                                const char *obtenido) {
    int ok = (obtenido != NULL) && (strcmp(esperado, obtenido) == 0);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %s\n", probadas, caso);
    printf("     esperado: %s\n", esperado);
    printf("     obtenido: %s\n", obtenido ? obtenido : "(NULL)");
    printf("     -> %s\n", ok ? "OK" : "FALLA");
}

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 24 - Saltos condicionales con cortocircuito ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { char l[8][64] = {{0}}; emitir_condicion(l, 8, "a", "b", '&', "L9"); verificar_str_largo("a && b (primer salto)", "IFFALSE a GOTO L9", l[0]); }
    { char l[8][64] = {{0}}; emitir_condicion(l, 8, "a", "b", '&', "L9"); verificar_str_largo("a && b (segundo salto)", "IFFALSE b GOTO L9", l[1]); }
    { char l[8][64] = {{0}}; int n = emitir_condicion(l, 8, "a", "b", '&', "L9"); verificar_int("a && b (dos saltos)", (long)(2), (long)(n)); }
    { char l[8][64] = {{0}}; emitir_condicion(l, 8, "a", "b", '|', "L9"); verificar_str_largo("a || b (primer salto)", "IFTRUE a GOTO L9", l[0]); }
    { char l[8][64] = {{0}}; emitir_condicion(l, 8, "a", "b", '|', "L9"); verificar_str_largo("a || b (segundo salto)", "IFTRUE b GOTO L9", l[1]); }
    { char l[4][64] = {{0}}; emitir_condicion(l, 4, "a", NULL, '&', "L0"); verificar_str_largo("condicion simple (b NULL)", "IFFALSE a GOTO L0", l[0]); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion24.md" "\n");
    return 0;
}
````
